```python
import pandas as pd
import random
```

```python
company_names = ["Acme Defense", "Global Parts", "Northstar Compents", "Sentinel Systems","Vanguard Industries", "Titan Manufacturing","Apex Electronics", "Ironclad Supply","Falcon Aerospace","Summit Machining"]
countries = ["USA","China","Germany","Taiwan","South Korea","Japan","Mexico","India"]
industries =["Electronics","Machined Parts","Software","Raw Materials","Circuit Boards","Optics"]
```

```python
suppliers = []
for i in range(40) :
    supplier = {
        "supplier_id" : i + 1,
        "name" : random.choice(company_names) + " " + str(i+ 1),
        "country" : random.choice(countries),
        "industry" : random.choice(industries),
        "risk_score" : round(random.uniform(1,10),1),
        "is_foreign_owned" : random.choice([True,False]),
        "single_source" : random.choice([True, False])
    }
    suppliers.append(supplier)

print(suppliers[0])
print(suppliers[1])
```

```python
df = pd.DataFrame(suppliers)
df.head()
```

```python
df . to_csv("suppliers.csv", index=False)
```

```python
import pandas as pd

df = pd.read_csv("suppliers.csv")
df.head()
```

```python
def calculate_foci_risk(row):
    score = row["risk_score"]

    if row["is_foreign_owned"]:
        score += 2

    if row["single_source"]:
        score +=1

    if score >=8:
        return "HIGH RISK"
    elif score >=5:
        return "MEDIUM RISK"
    else:
        return "LOW RISK"
```

```python
df["foci_risk"] = df.apply(calculate_foci_risk, axis=1)
df.head(10)
```

```python
df["foci_risk"]. value_counts()
```

```python
df.to_csv("suppliers_scored.csv", index=False)
```
